```python
import pandas as pd
data=pd.read_csv('todays_data.csv')
# Computes basic statistics like count, mean, std, min, max, and quartiles
data.describe()
```

```python
# Group by 'Academic Level'
grouped_data = data.groupby('Academic Level')

# Display the result
print(grouped_data)


```

```python
# Count of rows per group
group_counts = data['Academic Level'].value_counts()
print(group_counts)
```
