```python
import pandas as pd

# Replace 'your_file.csv' with the actual file path
df = pd.read_csv('todays_data.csv')

# Now try to get the value counts
counts = df['Academic Level'].value_counts()

# Print the counts
print(counts)





```

```python
import matplotlib.pyplot as plt

# Clean the index (for better formatting in the plot)
counts.index = counts.index.str.strip().str.title()  # Capitalize each label and remove extra spaces

# Plot the line graph
plt.plot(counts.index, counts.values, marker='o')
plt.xlabel('Academic Level')
plt.ylabel('Count')
plt.title('Count per Academic Level')
plt.xticks(rotation=45)  # Rotate labels if they overlap
plt.grid(True)
plt.show()
```
